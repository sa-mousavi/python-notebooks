```python
import matplotlib.pyplot as plt
categories = ['Group A', 'Group B', 'Group C', 'Group D']
values = [23, 45, 56, 32]

# رنگ متناظر با هر دسته، به همان ترتیب
colors = ['steelblue', 'forestgreen', 'crimson', 'darkorange']

fig, ax = plt.subplots(figsize=(7, 4.5))

bars = ax.bar(
    categories,
    values,
    width=0.6,
    color=colors,
    edgecolor='black',
    linewidth=1.1
)

ax.bar_label(bars, fontsize=12, padding=3)

ax.set_ylim(0, max(values) * 1.15)
ax.set_xlabel('Categories', fontsize=14, labelpad=8)
ax.set_ylabel('Values', fontsize=14, labelpad=10)
ax.set_title('Bar Chart with Individual Bar Colors', fontsize=16, pad=12)
ax.tick_params(axis='both', labelsize=14, pad=5) 

plt.savefig('./image/barcolor.pdf', bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# تنظیم فونت ریاضی مشابه کتابچه
plt.rcParams['mathtext.fontset'] = 'cm'

categories = ['A', 'B', 'C', 'D', 'E']
x = np.arange(len(categories))

values1 = [5, 7, 3, 4, 6]
values2 = [6, 6, 4, 5, 7]
values3 = [5, 6, 5, 4, 6]

bar_width = 0.25  # پهنای هر ستون

fig, ax = plt.subplots(figsize=(8, 5))

# رسم متقارن ستون‌ها نسبت به مرکز هر دسته
rects1 = ax.bar(x - bar_width, values1, width=bar_width, label='Series 1', color='steelblue')
rects2 = ax.bar(x,             values2, width=bar_width, label='Series 2', color='forestgreen')
rects3 = ax.bar(x + bar_width, values3, width=bar_width, label='Series 3', color='crimson')

# تنظیم عنوان و محورها
ax.set_title('A Multiseries Vertical Bar Chart', fontsize=18, pad=15)
ax.set_xlabel('Categories', fontsize=14, labelpad=10)
ax.set_ylabel('Values', fontsize=14, labelpad=10)

ax.set_xticks(x)
ax.set_xticklabels(categories)
ax.set_ylim(0, 9)
ax.tick_params(axis='both', labelsize=14, pad=5) 

# افزودن راهنما و خطوط شبکه افقی
ax.legend(frameon=True, fontsize=12, loc=2)
ax.grid(axis='y', linestyle='--', alpha=0.75)

plt.savefig('./image/barmulv.pdf', bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

months = ['Jan15', 'Feb15', 'Mar15', 'Apr15']
x = np.arange(len(months))

series1 = np.array([3, 4, 5, 3])
series2 = np.array([1, 2, 2, 5])
series3 = np.array([2, 3, 3, 4])

bar_width = 0.5  # پهنای مناسب و استاندارد برای ستون‌ها

fig, ax = plt.subplots(figsize=(7, 5))

# لایه اول (پایه)
p1 = ax.bar(x, series1, width=bar_width, label='Series 1', color='steelblue')

# لایه دوم (روی لایه اول)
p2 = ax.bar(x, series2, width=bar_width, bottom=series1, label='Series 2', color='forestgreen')

# لایه سوم (روی مجموع دو لایه اول)
p3 = ax.bar(x, series3, width=bar_width, bottom=series1 + series2, label='Series 3', color='crimson')

# تنظیم عنوان و محورها
ax.set_title('Stacked Bar Chart', fontsize=18, pad=15)
ax.set_xlabel('Months', fontsize=14, labelpad=10)
ax.set_ylabel('Values', fontsize=14, labelpad=10)

# تیک‌ها دقیقاً در مرکز هر ستون
ax.set_xticks(x)
ax.set_xticklabels(months)
ax.set_ylim(0, 15)

# راهنما و شبکه افقی
ax.legend(frameon=True, fontsize=12, loc=2)
ax.grid(axis='y', linestyle='--', alpha=0.5)
ax.tick_params(axis='both', labelsize=14, pad=5) 

plt.savefig('./image/barstackev.pdf', bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

months = ['Jan15', 'Feb15', 'Mar15', 'Apr15']
y = np.arange(len(months))

series1 = np.array([3, 4, 5, 3])
series2 = np.array([1, 2, 2, 5])
series3 = np.array([2, 3, 3, 4])

bar_height = 0.5  # ضخامت میله‌ها

fig, ax = plt.subplots(figsize=(7, 5))

# لایه اول (شروع از مبدا x=0)
p1 = ax.barh(y, series1, height=bar_height, label='Series 1', color='steelblue')

# لایه دوم (شروع از انتهای لایه اول با آرگومان left)
p2 = ax.barh(y, series2, height=bar_height, left=series1, label='Series 2', color='forestgreen')

# لایه سوم (شروع از انتهای لایه‌های اول و دوم)
p3 = ax.barh(y, series3, height=bar_height, left=series1 + series2, label='Series 3', color='crimson')

# درج عدد در مرکز هر بخش (اختیاری و خوانا)
ax.bar_label(p1, labels=['S1']*len(y), label_type='center', color='white', fontweight='bold')
ax.bar_label(p2, labels=['S2']*len(y), label_type='center', color='white', fontweight='bold')
ax.bar_label(p3, labels=['S3']*len(y), label_type='center', color='white', fontweight='bold')


# تنظیم عنوان و محورها
ax.set_title('Horizontal Stacked Bar Chart', fontsize=18, pad=15)
ax.set_xlabel('Values', fontsize=14, labelpad=10)
ax.set_ylabel('Months', fontsize=14, labelpad=10)

# تیک‌ها دقیقاً در امتداد میله‌ها
ax.set_yticks(y)
ax.set_yticklabels(months)
ax.set_xlim(0, 15)
ax.tick_params(axis='both', labelsize=14, pad=5) 

# راهنما و شبکه عمودی
ax.legend(frameon=True, fontsize=12, loc='lower right')
ax.grid(axis='x', linestyle='--', alpha=0.5)

plt.savefig('./image/barstackeh.pdf', bbox_inches='tight')
plt.show()
```
