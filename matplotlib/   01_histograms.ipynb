```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['mathtext.fontset'] = 'cm'
plt.rcParams['hatch.linewidth'] = 1.8  # رفع درزهای سفید در خروجی PDF

# ۱. تولید داده‌ها
np.random.seed(42)
data = np.random.normal(loc=100, scale=15, size=200)

# ۲. ایجاد بوم شیءگرا
fig, ax = plt.subplots(figsize=(8, 5))

# ۳. رسم هیستوگرام (استفاده از رنگ آبی فولادی: steelblue)
n, bins, patches = ax.hist(data, bins=8, edgecolor='black', linewidth=1.2, color='steelblue')

# ۴. اعمال ویژگی‌های ظاهری با نام‌های استاندارد رنگ
# ستون سوم: قرمز زرشکی
patches[2].set_facecolor('crimson')

# ستون چهارم: سبز جنگلی به همراه هاشور
patches[3].set_facecolor('forestgreen')
patches[3].set_hatch('//')

# ستون پنجم: تغییر شفافیت (alpha)
patches[4].set_alpha(0.35)

# ۵. درج برچسب مقادیر بالای ستون‌ها
ax.bar_label(patches, fmt='%.0f', fontsize=12, padding=3)

# ۶. افزایش دامنه محور عمودی برای جلوگیری از تداخل برچسب‌ها با حاشیه بالا
ax.set_ylim(0, max(n) * 1.2)

# ۷. تنظیمات محورها و عناوین
ax.set_xlabel(r'$X$', fontsize=14, labelpad=10)
ax.set_ylabel('Frequency', fontsize=14, labelpad=10)
ax.set_title('Histogram with Custom Patches and Bar Labels', fontsize=16, pad=10)
ax.tick_params(axis='both', labelsize=14, pad=5) 
plt.savefig('./image/hispatch.pdf', bbox_inches='tight')
plt.show()
```
